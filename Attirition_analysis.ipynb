## 7. Statistical Outlier Analysis Using IQR

The Interquartile Range (IQR) method is used to identify potential statistical outliers in selected numerical columns.

### IQR Method

- **Q1:** 25th percentile.
- **Q2:** 50th percentile (median).
- **Q3:** 75th percentile.
- **IQR:** Q3 − Q1.

The potential outlier boundaries are calculated as:

- Lower Bound = Q1 − 1.5 × IQR
- Upper Bound = Q3 + 1.5 × IQR

Values below the lower bound or above the upper bound are flagged as potential statistical outliers.

### Important Interpretation

An IQR outlier is not automatically an incorrect or invalid record.

The method identifies values that are relatively distant from the middle 50% of the data. Each flagged value must be investigated using relevant employee attributes and business context before deciding whether it should be retained, corrected or excluded.

No record will be removed solely because it falls outside the IQR boundaries.

### 7.1 Reusable IQR Function
A reusable function is created to calculate Q1, Q3, IQR, lower and upper boundaries, and the number of potential outliers for a selected numerical column.

The function returns the records identified as potential outliers so they can be investigated separately.

```python
#19
#IQR analysis
def check_iqr(data, column):

    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)

    outliers = data[
        (data[column] < lower_bound) |
        (data[column] > upper_bound)
    ]

    print(f"\nColumn: {column}")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Potential Outliers:", len(outliers))

    return outliers
```

## 8. Business Rule Validation

Statistical outlier detection identifies unusual values based on the distribution of the data. Business-rule validation checks whether records are logically consistent with the meaning of the employee attributes.

These checks help identify values that may be invalid even when they are not statistical outliers.

The following relationships are examined:

1. Age should comply with the applicable employee eligibility policy.
2. MonthlyIncome should be greater than zero.
3. TotalWorkingYears should be greater than or equal to YearsAtCompany, assuming both fields use compatible definitions.
4. YearsInCurrentRole should not exceed YearsAtCompany.
5. YearsWithCurrManager should not exceed YearsAtCompany.
6. EmployeeNumber should be unique.

**Important:** Business rules must be supported by the dataset definitions or an applicable organisational policy. A failed check identifies a record for investigation; it does not automatically prove that the record is incorrect.

```python
#27
# BUSINESS RULE VALIDATION

# Rule 1: Age validation
invalid_age = df[
    (df["Age"] < 18) |
    (df["Age"] > 65)                  ##used on bases of general criteria not on specific business policy
]

# Rule 2: Monthly income must be positive
invalid_salary = df[
    df["MonthlyIncome"] <= 0
]

# Rule 3: Total experience vs company tenure
invalid_experience = df[
    df["TotalWorkingYears"] < df["YearsAtCompany"]
]

# Rule 4: Current role tenure
invalid_role = df[
    df["YearsInCurrentRole"] > df["YearsAtCompany"]
]

# Rule 5: Manager tenure
invalid_manager = df[
    df["YearsWithCurrManager"] > df["YearsAtCompany"]
]

# Rule 6: Employee ID uniqueness
duplicate_ids = df[
    df["EmployeeNumber"].duplicated(keep=False)
]

# Display results
print("Invalid Age Records:", len(invalid_age))
print("Invalid Salary Records:", len(invalid_salary))
print("Invalid Experience Records:", len(invalid_experience))
print("Invalid Role Tenure Records:", len(invalid_role))
print("Invalid Manager Tenure Records:", len(invalid_manager))
print("Duplicate Employee IDs:", len(duplicate_ids))
```
